## Q1. Explain the roles of the Driver, Cluster Manager, and Executor in a Spark application.

- **Driver:** The process that runs the `main()` of the Spark application. It creates the `SparkSession`/`SparkContext`, builds the logical plan (DAG) from the code, splits it into stages and tasks, and coordinates the whole job. It also collects results back (e.g. on `.collect()`).
- **Cluster Manager:** The external service (YARN, Kubernetes, Mesos, or Spark's own Standalone manager) responsible for allocating resources (CPU, memory) across the cluster. It launches Executors on worker nodes on the Driver's request and tracks their health.
- **Executor:** A JVM process running on a worker node that actually executes the tasks assigned by the Driver. Each executor holds a slice of the data (partitions) in memory/disk, runs the transformations, and reports status/results back to the Driver. Executors live for the duration of the application (unless dynamic allocation is used).

In short: **Driver = brain/coordinator, Cluster Manager = resource allocator, Executors = workers that do the actual computation.**

## Q2. How does Spark's Lazy Evaluation strategy improve performance when chain-processing large datasets?

Spark doesn't execute transformations (`select`, `filter`, `withColumn`, etc.) the moment they're called. Instead it just records them as a **logical plan (DAG)**. Execution only happens when an **action** (`show()`, `collect()`, `write()`, `count()`) is triggered.

This improves performance because:
1. **Optimization before execution** — the Catalyst optimizer can look at the *entire* chain of transformations at once and reorder/combine them (e.g. push filters down before reads, combine multiple `.filter()` calls, prune unused columns) instead of optimizing step by step.
2. **Avoids unnecessary work** — intermediate results are never materialized unless required, so Spark skips computing data that never contributes to the final action.
3. **Better resource planning** — since the full DAG is known upfront, Spark can plan stages, partitions, and shuffles more efficiently rather than reacting to each command in isolation.

In short, laziness lets Spark compile a chain of steps into one optimized physical plan instead of running each transformation naively one at a time.

## Q4. Difference between CSV and Parquet in terms of storage (row-based vs columnar) and why it matters for performance.

| | CSV | Parquet |
|---|---|---|
| Storage layout | **Row-based** — each row is stored together | **Columnar** — each column is stored together |
| Schema | Not stored; must be inferred/guessed every read | Schema is embedded in the file (self-describing) |
| Compression | Poor (plain text) | Excellent (column values are similar, compress well) |
| Reading specific columns | Must read the whole row even if you need 2 of 20 columns | Can read only the needed columns (**column pruning**) |
| Filtering | No pushdown; all rows scanned | Supports **predicate pushdown** — skips row groups that can't match |

Because Parquet is columnar, compressed, and supports column pruning + predicate pushdown, it's much faster and cheaper (less I/O, less memory) for analytical workloads on large datasets — which is why it's preferred over CSV for big data pipelines. CSV is mainly useful for interoperability/human-readability, not performance.

## Q7. How does Spark use the Lineage Graph (DAG) to provide fault tolerance if a worker node fails?

Spark doesn't replicate data for fault tolerance the way HDFS replicates blocks. Instead, every RDD/DataFrame remembers the **lineage** — the sequence of transformations that produced it from the original source data (a DAG of parent → child dependencies).

If a worker node (and the partitions/executors on it) fails mid-job, Spark doesn't need to restart the whole application. It looks at the lineage graph, finds which partitions were lost, and **recomputes only those specific partitions** by re-running the recorded transformations from the last available data (source file or a cached/checkpointed stage). This is far cheaper than full replication and is the core of Spark's resilience (the "R" in RDD — Resilient Distributed Dataset).

## Q9. Explain the concept of Predicate Pushdown in Parquet and how it affects the amount of data loaded into memory.

Predicate Pushdown means Spark "pushes" a filter condition (the `WHERE`/`.filter()` predicate) down to the **file-reading layer** instead of applying it only after loading all the data into memory.

Parquet files store data in **row groups**, and each row group keeps min/max statistics for every column. When a filter is pushed down, Spark checks those statistics first and **completely skips reading row groups** that can't possibly satisfy the condition (e.g. if the filter is `price > 5000` and a row group's max price is 3000, that whole row group is skipped from disk I/O).

Effect: much less data is physically read from disk and loaded into memory/network, because irrelevant chunks are eliminated before they're even deserialized — resulting in faster queries and lower memory pressure, especially on large datasets. CSV can't do this since it has no per-block statistics or columnar layout.

## Q11. Difference between Transformations and Actions? (with two examples each)

- **Transformations** are **lazy** operations that define *how* to derive a new DataFrame/RDD from an existing one. They just build up the DAG and return a new DataFrame — nothing actually runs.
  - Examples: `select()`, `filter()`, `withColumn()`, `groupBy()`, `join()`

- **Actions** are operations that **trigger execution** of the DAG and return a result to the Driver (or write output).
  - Examples: `show()`, `count()`, `collect()`, `write.csv()`

Rule of thumb: if it returns a new DataFrame → transformation; if it returns a concrete value/output or writes data → action.

```
# Q11 - demonstration
# Transformations (lazy, nothing executes yet)
transformed = df.filter(col("region") == "North").select("order_id", "region", "amount")

# Actions (trigger the actual execution)
print("Row count (action):", transformed.count())
transformed.show(5)  # action
```

## Q13. In Spark Architecture, what is the difference between Client Mode and Cluster Mode?

- **Client Mode:** The Driver runs on the **machine that submitted the job** (e.g. your laptop, edge node, or notebook), while Executors run on the cluster. Good for interactive use (notebooks, debugging, `spark-shell`) since you see logs/output directly, but if the client machine disconnects or crashes, the whole application dies. Also, more network traffic flows between the client and the cluster for driver communication.

- **Cluster Mode:** The Driver itself is launched **inside the cluster** (on one of the worker/cluster nodes), managed by the cluster manager. Better for production/scheduled jobs — the client can submit and disconnect, and the job keeps running independent of the client machine's availability. Logs are collected via the cluster manager (e.g. YARN UI) instead of the local terminal.

In short: **Client mode = Driver stays on your machine (interactive); Cluster mode = Driver runs inside the cluster (production, resilient to client disconnects).**

## Q15. When exploring a dataset, why is it safer to use `.show(5)` instead of `.collect()` on a multi-terabyte dataset?

- `.show(5)` only computes and displays a **small preview** (default 20 rows, or 5 if specified) — Spark is smart enough to only process the minimum data needed to produce those rows, and the result **stays distributed on the executors**, never pulled fully into the Driver's memory.
- `.collect()` forces Spark to gather **every single row of the entire dataset** back to the Driver's memory as a local Python/Scala list. On a multi-terabyte dataset, this will almost certainly **crash the Driver with an OutOfMemory error**, since the Driver typically has far less memory than the whole cluster combined.

So for exploration/debugging, `.show()` (or `.take(n)`) is safe because it's bounded and distributed, while `.collect()` is only safe on small, already-aggregated/filtered results that are known to fit in Driver memory.

## Performance & Architecture Insights

- **Lazy evaluation + DAG**: Spark builds an optimized execution plan before running anything, avoiding redundant computation.
- **Parquet vs CSV**: Parquet's columnar layout + predicate pushdown + column pruning drastically cut down I/O compared to row-based CSV, especially as data scales to GBs/TBs.
- **Fault tolerance** comes from lineage (recompute lost partitions) rather than data replication, keeping Spark efficient.
- **Client vs Cluster mode** matters for where jobs are best submitted from — cluster mode is preferred for production pipelines.
- **Avoid `.collect()` on large data** — always prefer `.show()`, `.take()`, or writing to a sink, and only collect small, aggregated results.
- Filtering **before** wide operations (joins/groupBy) and pushing filters down to the file format (Parquet) reduces shuffle and memory pressure significantly.
